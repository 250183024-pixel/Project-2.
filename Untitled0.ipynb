```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

sns.set()
```

```python
df = pd.read_csv('ecommerce_orders.csv')
```

```python
plt.figure()

for seg in df['Segment'].unique():
    data = df[df['Segment'] == seg]
    sns.kdeplot(data['Order_Value'], label=seg)

plt.xscale('log')
plt.title('Order Value Distribution')
plt.xlabel('Order Value')
plt.ylabel('Density')
plt.legend()

plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12,4))

for ax, ch in zip(axes, df['Channel'].unique()):
    data = df[df['Channel'] == ch]#Keep only rows for this channel

    for seg in data['Segment'].unique():
        grp = data[data['Segment'] == seg]#filters

        ax.scatter(grp['Discount'], grp['Profit_Margin'])

        # trend line
        m, b = np.polyfit(grp['Discount'], grp['Profit_Margin'], 1)
        x = np.linspace(grp['Discount'].min(), grp['Discount'].max(), 50)
        ax.plot(x, m*x + b)#also formula

    ax.set_title(ch)
    ax.set_xlabel('Discount')
    ax.set_ylabel('Profit')

plt.tight_layout()
plt.show()
```

```python
pivot_rev = df.pivot_table(values='Revenue', index='Region', columns='Segment')
pivot_margin = df.pivot_table(values='Profit_Margin', index='Region', columns='Segment')

plt.figure()
sns.heatmap(pivot_rev, annot=True)
plt.title('Average Revenue')
plt.show()

plt.figure()
sns.heatmap(pivot_margin, annot=True)
plt.title('Average Profit Margin')
plt.show()
```
